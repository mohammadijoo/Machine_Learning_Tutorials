# فصل ۸ — یادگیری مبتنی بر نمونه
## درس ۳: معیارهای فاصله و وزن‌دهی

K-Nearest Neighbors makes its decision from two ideas: which observations count as nearby, and how strongly each nearby observation should vote. A distance metric defines nearby. A weighting rule defines influence. These choices are not minor implementation switches. They express assumptions about the geometry of the data and can change a classifier even when the training rows and the value of $k$ remain fixed.

This lesson studies Euclidean, Manhattan, Chebyshev, and general Minkowski distances, then compares uniform and distance-weighted votes. We will see why feature scaling is necessary, why categorical variables need an explicit representation, why high-dimensional distance becomes difficult to interpret, and why no metric is best without considering the data-generating process. A metric is useful when its notion of similarity agrees with the problem, not simply because it is available in a library.

The main example uses drug200.csv from this repository. It contains numeric patient attributes together with categorical Sex, BP, and Cholesterol variables, and Drug is the class label. This mixed structure lets us demonstrate scaling plus one-hot encoding before computing distances. A second example uses stars.csv and focuses on numeric astronomical measurements with very different ranges. Both examples are evaluated with validation rather than intuition alone.

### Learning goals

By the end of this lesson, you should be able to:

1. State the difference between a distance metric and a neighbor-voting weight.
2. Compute and interpret Euclidean, Manhattan, Chebyshev, and Minkowski distances.
3. Explain why scaling changes a distance-based model and apply it safely in a Pipeline.
4. Prepare numeric and categorical columns for KNN without inventing an artificial category order.
5. Compare distance metrics with stratified cross-validation and an appropriate score.
6. Compare uniform and distance-weighted neighbor votes.
7. Inspect the training examples and distances behind a particular KNN prediction.
8. Recognize limitations involving irrelevant features, sparse spaces, outliers, and the curse of dimensionality.

The most important habit is to treat distance design as modeling. Before asking which metric is highest-scoring, ask what a coordinate difference means, whether every coordinate is trustworthy, and whether the representation preserves the domain concept of similarity.

## 1. The geometry behind nearest neighbors

For two feature vectors $x=(x_1,\ldots,x_p)$ and $z=(z_1,\ldots,z_p)$, the Minkowski family is

$$
d_q(x,z)=\left(\sum_{j=1}^{p}|x_j-z_j|^q\right)^{1/q}.
$$

With $q=2$, this is Euclidean distance, the straight-line distance familiar from geometry. With $q=1$, it is Manhattan distance, which adds absolute coordinate changes. As $q$ becomes very large, the distance approaches Chebyshev distance, $d_\infty(x,z)=\max_j |x_j-z_j|$, determined by the largest coordinate difference. Each metric creates a different shape of equal-distance neighborhoods: circles or spheres for Euclidean distance, diamonds for Manhattan distance, and squares or cubes for Chebyshev distance.

Those shapes matter. A query near a diagonal cluster may have different nearest observations under Manhattan and Euclidean distance. Chebyshev distance is strict about the worst coordinate mismatch; it can make sense when one badly mismatched constraint should dominate similarity. Euclidean distance is often a good default for continuously measured, scaled variables, but it assumes squared coordinate differences are a sensible way to combine evidence. No formula can make an arbitrary feature representation meaningful.

## 2. What neighbor weighting changes

After a metric chooses the $k$ closest observations, KNN must aggregate their labels. Uniform weighting gives each selected neighbor one vote. Distance weighting gives closer neighbors more influence, commonly through a weight such as

$$
w_i=\frac{1}{d(x_*,x_i)+\epsilon}.
$$

The predicted class is the one with the largest weighted local total. Uniform voting is robust when a neighborhood is reasonably homogeneous and distances are noisy. Distance weighting can help when a query sits very close to one informative observation but several farther observations belong to another class. It can also amplify a noisy nearest point, especially with small $k$ or poor feature scaling.

Metric choice and weighting interact. Changing the metric changes the distances and possibly the membership of the neighborhood. Changing weights changes the influence of members without changing membership. It is therefore sensible to compare combinations inside the same validation protocol rather than declaring one option universally superior.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.compose import ColumnTransformer
from sklearn.metrics import ConfusionMatrixDisplay, accuracy_score, classification_report, f1_score, make_scorer
from sklearn.model_selection import StratifiedKFold, cross_val_score, train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, StandardScaler

RANDOM_STATE = 42
sns.set_theme(style="whitegrid", context="notebook")
macro_f1 = make_scorer(f1_score, average="macro", zero_division=0, pos_label=None)
```

## 3. Inspect mixed feature types in drug200.csv

The first rows show that Age and Na_to_K are numeric measurements, while Sex, BP, and Cholesterol are categorical. Drug is the target. We must not assign an arbitrary numeric order to categories, such as LOW=0, NORMAL=1, HIGH=2, and then treat the gaps as physical distances. That representation would claim, for example, that NORMAL is exactly halfway between LOW and HIGH, which may not be a valid clinical statement.

One-hot encoding represents each category with a binary indicator. After standardizing the numeric columns, a distance can combine numeric differences and category mismatches. This is a modeling approximation: one category mismatch contributes a defined amount to the encoded distance. It is transparent and often effective, but it should be reviewed with domain knowledge. Different categorical similarity rules can be appropriate in other applications.

```python
drug_path = Path("../../../Datasets/Classification/drug200.csv")
drug_df = pd.read_csv(drug_path)
drug_target = "Drug"
drug_X = drug_df.drop(columns=drug_target)
drug_y = drug_df[drug_target]
numeric_columns = drug_X.select_dtypes(include="number").columns.tolist()
categorical_columns = drug_X.select_dtypes(exclude="number").columns.tolist()

display(drug_df.head())
display(drug_df.dtypes.rename("dtype").to_frame())
display(pd.DataFrame({"numeric_columns": pd.Series(numeric_columns), "categorical_columns": pd.Series(categorical_columns)}))
display(drug_y.value_counts().rename("count").to_frame())
```

## 4. Build a distance-ready representation

The preprocessor below standardizes numeric columns and one-hot encodes categorical columns. It is inside the pipeline, so each cross-validation training fold learns its own scaling values and category layout. handle_unknown="ignore" makes prediction robust when a later row contains a category not observed in a particular training fold. sparse_output=False produces a dense representation, which lets us compare the dense distance metrics consistently.

This representation has a consequence worth noticing. Two patients with different BP categories have different indicator patterns, and their categorical contribution to distance is determined by those mismatches. If certain categories are more similar than others, simple one-hot encoding cannot express that nuance. A specialized domain distance, learned embedding, or a different algorithm may be better. KNN succeeds only when its geometric representation reflects useful similarity.

```python
drug_preprocessor = ColumnTransformer([("numeric", StandardScaler(), numeric_columns), ("categorical", OneHotEncoder(handle_unknown="ignore", sparse_output=False), categorical_columns)], sparse_threshold=0)
drug_X_train, drug_X_test, drug_y_train, drug_y_test = train_test_split(drug_X, drug_y, test_size=0.25, stratify=drug_y, random_state=RANDOM_STATE)
drug_cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
prepared_training = drug_preprocessor.fit_transform(drug_X_train)
prepared_columns = drug_preprocessor.get_feature_names_out()
display(pd.DataFrame(prepared_training, columns=prepared_columns).head())
print(f"Original feature columns: {drug_X.shape[1]}")
print(f"Distance-ready feature columns: {prepared_training.shape[1]}")
```

## 5. Compare Euclidean, Manhattan, and Chebyshev distance

We compare three metrics with the same preprocessing, $k=7$, uniform votes, and five-fold stratified cross-validation on the training subset. Macro F1 gives each drug class equal weight in the comparison. The exercise does not prove a metric is globally best; it asks which distance representation is most useful for these data under this pipeline and score.

Euclidean distance accumulates squared differences and tends to emphasize larger coordinate gaps after scaling. Manhattan distance adds absolute differences and can be less dominated by a single large discrepancy. Chebyshev distance looks only at the largest discrepancy, which is appropriate only when the worst mismatch has special meaning. The result table and chart provide empirical evidence, while the interpretation requires domain judgment.

```python
metric_settings = [("Euclidean", {"metric": "minkowski", "p": 2}), ("Manhattan", {"metric": "minkowski", "p": 1}), ("Chebyshev", {"metric": "chebyshev"})]
metric_rows = []
for metric_name, metric_parameters in metric_settings:
    metric_model = make_pipeline(drug_preprocessor, KNeighborsClassifier(n_neighbors=7, weights="uniform", **metric_parameters))
    metric_scores = cross_val_score(metric_model, drug_X_train, drug_y_train, cv=drug_cv, scoring=macro_f1, error_score="raise")
    metric_rows.append({"metric": metric_name, "mean_macro_f1": metric_scores.mean(), "std_macro_f1": metric_scores.std()})
metric_results = pd.DataFrame(metric_rows).sort_values("mean_macro_f1", ascending=False)
display(metric_results.round(3))

fig, ax = plt.subplots(figsize=(8, 5))
ax.bar(metric_results["metric"], metric_results["mean_macro_f1"], yerr=metric_results["std_macro_f1"], capsize=5, color=["tab:blue", "tab:orange", "tab:green"])
ax.set_ylim(0, 1.05)
ax.set_xlabel("Distance metric")
ax.set_ylabel("Cross-validation macro F1")
ax.set_title("Metric comparison for drug classification")
plt.show()
```

## 6. See how the metrics shape neighborhoods

A two-dimensional toy view makes the geometry visible. Around the origin, Euclidean equal-distance contours are circular, Manhattan contours are diamond shaped, and Chebyshev contours are square. In more dimensions the shapes become spheres, cross-polytopes, and hypercubes, but the same principle remains: the chosen metric determines which observations enter a neighborhood first.

This visualization is not a claim that real data are two-dimensional. It is a way to reason about the behavior of a metric. If domain similarity is constrained by the single largest mismatch, Chebyshev-like behavior can be justified. If movement along coordinates has an additive cost, Manhattan distance can be natural. If independent, scaled measurement noise is roughly isotropic, Euclidean distance is often sensible. The data and the measurement process should guide the choice.

```python
grid_values = np.linspace(-3, 3, 300)
grid_x, grid_y = np.meshgrid(grid_values, grid_values)
euclidean_distance = np.sqrt(grid_x**2 + grid_y**2)
manhattan_distance = np.abs(grid_x) + np.abs(grid_y)
chebyshev_distance = np.maximum(np.abs(grid_x), np.abs(grid_y))

fig, axes = plt.subplots(1, 3, figsize=(15, 4.5), sharex=True, sharey=True)
for axis, distance_grid, title in zip(axes, [euclidean_distance, manhattan_distance, chebyshev_distance], ["Euclidean distance", "Manhattan distance", "Chebyshev distance"]):
    axis.contour(grid_x, grid_y, distance_grid, levels=[0.5, 1.0, 1.5, 2.0, 2.5], cmap="viridis")
    axis.scatter([0], [0], color="black", marker="x", s=60)
    axis.set_title(title)
    axis.set_xlabel("Feature 1 difference")
axes[0].set_ylabel("Feature 2 difference")
plt.tight_layout()
plt.show()
```

## 7. Compare uniform and distance-weighted votes

Next, we hold the metric fixed at the strongest candidate from the prior comparison and compare two voting rules. The metric decides the ordering of neighbors. Uniform voting treats every one of the seven selected rows equally. Distance weighting increases the influence of closer rows. This can be helpful near a class boundary, but it also makes tiny distances more influential.

A validation result should decide whether the extra local sensitivity helps. It is tempting to assume that distance weighting must improve KNN because closer points seem more relevant. That intuition can fail when distance estimates are noisy, when a closest point is an outlier, or when the local region is sparsely sampled. The evidence below is specific to this training data and preprocessing design.

```python
best_metric_name = metric_results.iloc[0]["metric"]
best_metric_parameters = dict(metric_settings)[best_metric_name]
weight_rows = []
for weight_rule in ["uniform", "distance"]:
    weight_model = make_pipeline(drug_preprocessor, KNeighborsClassifier(n_neighbors=7, weights=weight_rule, **best_metric_parameters))
    weight_scores = cross_val_score(weight_model, drug_X_train, drug_y_train, cv=drug_cv, scoring=macro_f1, error_score="raise")
    weight_rows.append({"weights": weight_rule, "mean_macro_f1": weight_scores.mean(), "std_macro_f1": weight_scores.std()})
weight_results = pd.DataFrame(weight_rows)
display(weight_results.round(3))
```

## 8. Inspect one local prediction

KNN is often easier to audit than a model that hides all evidence in many fitted parameters. For one test patient, we can transform the features with the fitted preprocessor, find the selected training neighbors, view their original rows, and inspect their transformed-space distances. This is a local explanation of the model's rule, not a causal explanation of which treatment is appropriate.

A useful inspection asks several questions. Are the nearest patients genuinely similar in the raw features? Are the categorical matches sensible? Is a single extremely close neighbor dominating a distance-weighted vote? Are classes mixed, suggesting uncertainty? These questions can reveal representation mistakes that a cross-validation score alone may not expose.

```python
selected_weight_rule = weight_results.loc[weight_results["mean_macro_f1"].idxmax(), "weights"]
selected_drug_model = make_pipeline(drug_preprocessor, KNeighborsClassifier(n_neighbors=7, weights=selected_weight_rule, **best_metric_parameters))
selected_drug_model.fit(drug_X_train, drug_y_train)
query_patient = drug_X_test.iloc[[0]]
query_transformed = selected_drug_model.named_steps["columntransformer"].transform(query_patient)
neighbor_distances, neighbor_indices = selected_drug_model.named_steps["kneighborsclassifier"].kneighbors(query_transformed)
neighbor_rows = drug_X_train.iloc[neighbor_indices[0]].copy()
neighbor_rows["Drug"] = drug_y_train.iloc[neighbor_indices[0]].to_numpy()
neighbor_rows["distance"] = neighbor_distances[0]
prediction_probabilities = pd.DataFrame({"Drug": selected_drug_model.classes_, "probability": selected_drug_model.predict_proba(query_patient)[0]})
display(query_patient.assign(predicted_drug=selected_drug_model.predict(query_patient)))
display(neighbor_rows.sort_values("distance").round(3))
display(prediction_probabilities.sort_values("probability", ascending=False).round(3))
```

## 9. Final held-out evaluation for drug classification

The metric and weighting rule have been selected from training-fold evidence. We now evaluate the resulting pipeline once on drug_X_test. The classification report separates performance by drug class, and the confusion matrix shows which medications are mixed by the available features. A strong score demonstrates that the representation and local rule work for this held-out split; it does not turn the notebook into a clinical decision system.

In a real treatment setting, target labels, data provenance, patient safety, subgroup fairness, missingness, and decision costs would require much more scrutiny. This example is about the mechanics of a distance-based classifier and the discipline of validating its assumptions.

```python
drug_test_predictions = selected_drug_model.predict(drug_X_test)
drug_test_accuracy = accuracy_score(drug_y_test, drug_test_predictions)
drug_test_macro_f1 = f1_score(drug_y_test, drug_test_predictions, average="macro", zero_division=0)
display(pd.DataFrame({"metric": ["accuracy", "macro_f1"], "test_value": [drug_test_accuracy, drug_test_macro_f1]}).round(3))
display(pd.DataFrame(classification_report(drug_y_test, drug_test_predictions, output_dict=True, zero_division=0)).T.round(3))

fig, ax = plt.subplots(figsize=(8, 6))
ConfusionMatrixDisplay.from_predictions(drug_y_test, drug_test_predictions, cmap="Blues", colorbar=False, ax=ax)
ax.set_title("Held-out performance of the selected drug KNN model")
plt.show()
```

## 10. A second dataset with extreme numeric scales: stars.csv

The stars dataset contains Temperature, luminosity L, radius R, absolute magnitude A_M, and categorical descriptors. Its first rows show that numeric columns can have radically different ranges: luminosity and radius vary over orders of magnitude, while absolute magnitude has a much narrower range. For this metric comparison, we use the numeric columns only and apply log transforms to positive L and R before standardization. A log transform compresses extreme scale differences while preserving order.

This is a good reminder that standardization is not the only transformation choice. StandardScaler puts transformed columns on comparable variance scales, but highly skewed positive variables may still benefit from a log transform first. Transformations should be justified by the measurements and inspected, not added mechanically.

```python
stars_path = Path("../../../Datasets/Classification/stars.csv")
stars_df = pd.read_csv(stars_path)
stars_target = "Type"
stars_y = stars_df[stars_target]
stars_X = stars_df[["Temperature", "L", "R", "A_M"]].copy()
stars_X["log_L"] = np.log10(stars_X.pop("L"))
stars_X["log_R"] = np.log10(stars_X.pop("R"))

display(stars_df.head())
display(stars_df.dtypes.rename("dtype").to_frame())
display(stars_y.value_counts().rename("count").to_frame())
display(stars_X.describe().T.round(3))
```

## 11. Metric comparison on transformed star measurements

We repeat a compact validation comparison on the star features. All inputs in this example are numeric after the two log transforms, so the geometry is easier to interpret than the mixed drug representation. The target has multiple stellar types, and macro F1 again prevents a common class from dominating the evaluation.

The selected metric may differ from the drug result. That is expected. A metric is not a global brand choice; it is a statement about a particular feature space. Reusing a validation protocol is valuable because it lets us compare assumptions fairly, even when the eventual best configuration changes.

```python
stars_X_train, stars_X_test, stars_y_train, stars_y_test = train_test_split(stars_X, stars_y, test_size=0.25, stratify=stars_y, random_state=RANDOM_STATE)
stars_cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
stars_rows = []
for metric_name, metric_parameters in metric_settings:
    stars_model = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=7, weights="distance", **metric_parameters))
    stars_scores = cross_val_score(stars_model, stars_X_train, stars_y_train, cv=stars_cv, scoring=macro_f1, error_score="raise")
    stars_rows.append({"metric": metric_name, "mean_macro_f1": stars_scores.mean(), "std_macro_f1": stars_scores.std()})
stars_results = pd.DataFrame(stars_rows).sort_values("mean_macro_f1", ascending=False)
display(stars_results.round(3))

fig, ax = plt.subplots(figsize=(8, 5))
ax.bar(stars_results["metric"], stars_results["mean_macro_f1"], yerr=stars_results["std_macro_f1"], capsize=5, color=["tab:purple", "tab:pink", "tab:cyan"])
ax.set_ylim(0, 1.05)
ax.set_xlabel("Distance metric")
ax.set_ylabel("Cross-validation macro F1")
ax.set_title("Metric comparison for transformed star measurements")
plt.show()
```

## 12. Distance limitations and practical design rules

In low dimensions, nearest observations can be genuinely close. As dimensions increase, distances tend to concentrate: the nearest and farthest points become less distinguishable relative to their magnitude. This curse of dimensionality means adding many weak or irrelevant features can make every neighbor less meaningful. Scaling cannot repair irrelevant information. Feature selection, dimensionality reduction, and domain-aware representations can be more valuable than endlessly testing metrics.

Outliers create another challenge. Euclidean distance squares differences before summing, so large discrepancies can have substantial influence. Manhattan distance is often more tolerant of one large coordinate difference, but it is not immune to poor measurements. Categorical one-hot variables can also dominate if there are many high-cardinality columns. Consider whether categories should be grouped, whether rare categories need special handling, and whether a different model family better fits the data.

A sound distance workflow is therefore: inspect the raw columns, remove identifiers and leakage, choose transformations that respect measurement meaning, keep transformations inside a pipeline, compare a small justified set of metrics and weights with cross-validation, inspect local neighbors, and evaluate once on a held-out set.

### Summary

KNN does not have one built-in meaning of near. A distance metric creates the geometry of neighborhoods, and a weighting rule controls how evidence inside those neighborhoods is aggregated. Euclidean, Manhattan, and Chebyshev distance emphasize different coordinate patterns. Uniform and distance weighting express different beliefs about local influence. We used pipelines to scale numeric features and encode categories safely, compared choices by cross-validation, inspected actual neighbors, and evaluated selected models on unseen data. The right metric and weighting rule are the ones that match a well-designed representation and hold up under honest validation.
